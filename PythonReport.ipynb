다음과 같은 작업을 수행합니다.

- 리스트의 데이터 순회
- 조건에 맞는 데이터 찾기
- 조건을 만족하는 데이터 개수 계산

#### AI를 통해 출제한 과제 문제 입니다

과제 문제

문제: 다음 판매 데이터를 이용하여 판매 데이터를 분석하는 프로그램을 작성하세요.

sales = [3, 12, 7, 15, 2, 20, 8, 11, 5, 18]

다음 조건을 모두 만족해야 합니다.

1. sales 리스트를 생성합니다.
2. for문을 이용하여 리스트의 데이터를 하나씩 확인합니다.
3. 판매량이 10개 이상인 데이터를 찾습니다.
4. 조건을 만족하는 판매량을 출력합니다.
5. 조건을 만족하는 데이터의 개수를 계산합니다.
6. 조건을 만족하는 데이터의 합계를 계산합니다.
7. 최종 결과를 출력합니다

1. sales 리스트를 생성합니다.
2. for문을 이용하여 리스트의 데이터를 하나씩 확인합니다.

````python
sales = [3, 12, 7, 15, 2, 20, 8, 11, 5, 18] # 리스트 생성

for sale in sales:
    print(f"판매량: {sale}")
````

3. 판매량이 10개 이상인 데이터를 찾습니다.
4. 조건을 만족하는 판매량을 출력합니다.
5. 조건을 만족하는 데이터의 개수를 계산합니다.
6. 조건을 만족하는 데이터의 합계를 계산합니다.
7. 최종 결과를 출력합니다

````python
count_greater_than_10 = 0
sum_greater_than_10 = 0

for sale in sales:
    if sale > 10:
        print(f"판매량이 10보다 큰 값: {sale}")
        count_greater_than_10 += 1
        sum_greater_than_10 += sale

print(f"판매량이 10보다 큰 값의 개수: {count_greater_than_10}")
print(f"판매량이 10보다 큰 값의 합계: {sum_greater_than_10}")
````

### Part 2. 딕셔너리를 이용한 단어 빈도 분석

다음과 같은 문장을 대상으로 단어가 몇 번 등장하는지 분석합니다.

```
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""
```

다음 과정을 직접 구현합니다.

1. `split()`으로 문장을 단어 단위로 나눕니다.
2. 빈 딕셔너리를 만듭니다.
3. 단어를 하나씩 반복합니다.
4. 처음 등장한 단어는 딕셔너리에 `1`로 저장합니다.
5. 이미 존재하는 단어는 기존 값에 `1`을 증가시킵니다.
6. 전체 단어와 등장 횟수를 출력합니다.
7. 일정 횟수 이상 등장한 단어만 별도로 출력합니다.

````python
sentence="""
파이썬 공부는 재미있다
파이썬 공부는 어렵지만 재미있다
데이터 분석에도 파이썬을 사용한다
파이썬 공부를 계속하면 실력이 늘어난다
"""

split_sentences = sentence.split()
print(split_sentences)
print(type(split_sentences))
````

````python
dict_sentences = {}
for i in range(len(split_sentences)):
    if split_sentences[i] not in dict_sentences:
        dict_sentences[split_sentences[i]] = 1
    else:
        dict_sentences[split_sentences[i]] += 1

print(dict_sentences)

print("\n\n2번 이상 등장한 단어")
for key, value in dict_sentences.items():
    if value >= 2:
        print(f"'{key}'의 개수: {value}")
````

#### AI를 통해 출제한 과제 문제 입니다

예제 문제 — 단어 빈도 분석 함수 만들기

다음 문장이 있습니다.

sentence = "python data python analysis data python study"

문장에 포함된 단어가 각각 몇 번 등장하는지 계산하는 프로그램을 작성하세요.

````python
sentence = "python data python analysis data python study"

def count_words(sentence):
    words = sentence.split()
    word_count = {}
    for word in words:
        if word not in word_count:
            word_count[word] = 1
        else:
            word_count[word] += 1
    return word_count

word_counts = count_words(sentence)
print(word_counts)
````

### Part 4. 오류 확인 및 코드 검증

작성한 코드에 문제가 발생했을 때 바로 전체 코드를 다시 만들지 않고 오류 원인을 확인합니다.

다음 순서로 확인합니다.

```
오류 메시지 확인
→ 문제가 발생한 코드 위치 확인
→ 변수 값 확인
→ 자료형 확인
→ 코드 수정
→ 다시 실행
```

가능하다면 예상 가능한 오류에 대해 `try-except`를 사용해 처리해봅니다.

````python
# 두 수를 입력받아 나눗셈 기능을 하는 함수 호출을 통해 실행하는 프로그램 입니다.

# try - except를 통한 예외 처리 구현
# 숫자 입력을 받을 때 숫자가 아닌 경우에 대한 예외 처리
# 0으로 나누는 경우에 대한 예외 처리

def divide(a, b):
    # 두 숫자를 나누어 결과를 반환하세요.
    return a / b

try:
    a = int(input("첫 번째 숫자: "))
    b = int(input("두 번째 숫자: "))

    result = divide(a, b)

    print("나눗셈 결과:", result)

except ValueError:
    print("숫자만 입력할 수 있습니다.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
````

다음 기능 중 하나 이상을 추가해보세요.

- 가장 많이 등장한 단어 찾기
- 가장 적게 등장한 단어 찾기
- 특정 단어가 몇 번 등장했는지 검색하기
- 사용자가 입력한 기준 횟수 이상 등장한 단어만 출력하기
- 분석 결과를 파일로 저장하기

````python
sentence = "python data data data python analysis data python study"

def count_words(sentence):
    words = sentence.split()
    word_count = {}
    for word in words:
        if word not in word_count:
            word_count[word] = 1
        else:
            word_count[word] += 1
    return word_count

word_counts = count_words(sentence)
print(word_counts)

print("\n가장 많이 등장한 단어")
max_word = {}
max_count = 0
for key, value in word_counts.items():
    if value > max_count:
        max_count = value
        max_word = key

print(f"'{max_word}' - {max_count}번")

print("\n가장 적게 등장한 단어")
min_word = {}
# word_counts의 첫 번째 값으로 초기화
print(type(word_counts.values()), word_counts.values())
min_count = list(word_counts.values())[0]
for key, value in word_counts.items():
    if value < min_count:
        min_count = value
        min_word = key

print(f"'{min_word}' - {min_count}번")

print("\n특정 단어가 몇 번 등장 했는지 검색")
search_word = "python"
if search_word in word_counts:
    print(f"'{search_word}'은 문장에 {word_counts[search_word]}번 등장했습니다.")
else:
    print(f"'{search_word}'는 문장에 없습니다.")

print("\n사용자가 입력한 기준 횟수 이상 등장한 단어 검색")
count = int(input("기준 횟수 입력: "))
for word, word_count in word_counts.items():
    if word_count >= count:
        print(f"'{word}' - {word_count}번")

# 파일에 저장
# sentence = "python data data data python analysis data python study"
with open("result.txt", "w", encoding="utf-8") as file:
    file.write("문장 분석 결과\n")
    file.write(f"예문: {sentence}\n\n")

    file.write("가장 많이 등장한 단어\n")
    file.write(f"'{max_word}' - {max_count}번\n\n")

    file.write("가장 적게 등장한 단어\n")
    file.write(f"'{min_word}' - {min_count}번\n\n")

    file.write("특정 단어가 몇 번 등장 했는지 검색\n")
    search_word = "python"
    if search_word in word_counts:
        file.write(f"'{search_word}'은 문장에 {word_counts[search_word]}번 등장했습니다.\n\n")
    else:
        file.write(f"'{search_word}'는 문장에 없습니다.\n")

    file.write(f"사용자가 입력한 기준 횟수({count}회 이상) 등장한 단어 검색\n")
    
    for word, word_count in word_counts.items():
        if word_count >= count:
            file.write(f"'{word}' - {word_count}번\n")
````
