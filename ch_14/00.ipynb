```python
# 두 수(1~100)를 입력 반아서 사칙연산 결과를 반환하는 calc 라는 함수를 만드시오.
# 4개 결과값을 한 번에 반환(Tuple)
# 사용자가 입력한 값이 숫자가 아닌경우 오류 메세지 출력하고 프로그램 종료

def calc(num1, num2):
    result = []

    if (num1 >=1 and num1 <=100) and (num2 >=1 and num2 <=100):
        result.append(num1 + num2)
        result.append(num1 - num2)
        result.append(num1 * num2)
        result.append(num1 / num2)
    else:
        return "입력값이 범위(1-100)를 벗어났습니다."

    return result

def check_num(num1):
    if len(num1) == 0:
        return False
    
    for i in num1:
        if ord(i) < 48 or ord(i) > 57:
            return False
    return True

num1 = input("숫자1을 입력하세요 : ")
num2 = input("숫자2을 입력하세요 : ")

if check_num(num1) and check_num(num2):
    ret = calc(int(num1),int(num2))
    print(ret)
else:
    print("입력값이 숫자가 아닙니다.")
```
