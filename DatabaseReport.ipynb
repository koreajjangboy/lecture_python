# SQL & 데이터베이스 과제

### Part 1. PostgreSQL Schema와 Table 만들기

````python
create SCHEMA practice;

create TABLE practice.members (
    member_id SERIAL primary KEY,
    name VARCHAR(50) not NULL,
    email VARCHAR(100) unique not NULL,
    age INTEGER,
    joined_at DATE
);
````

### Part 4. 집계 함수를 이용한 회원 데이터 분석

다음 정보를 SQL로 조회합니다.

- 전체 회원 수
- 회원 평균 나이
- 가장 나이가 많은 회원의 나이
- 가장 나이가 어린 회원의 나이
- 25세 이상 회원 수

다음 함수를 활용합니다.

```
COUNT()
AVG()
MAX()
MIN()
```

**필수 역량**

기본 SQL 집계 함수

````python
select count(*) as "전체 회원수" from practice.members;
````

전체 회원수
4

````python
select avg(age) as "회원 평균 나이" from practice.members;
````

회원 평균 나이
28.5

````python
select max(age) as "가장 나이가 많은 회원의 나이", min(age) as "가장 나이가 어린 회원의 나이" from practice.members;
````

가장 나이가 많은 회원의 나이	가장 나이가 어린 회원의 나이
32	                        24

````python
select count(*) as "25세 이상 회원 수" from practice.members where age >= 25;
````

25세 이상 회원 수
3

### 도전 문제

다음 문제 중 하나 이상을 해결해보세요.

- 연령대별 회원 수를 조회하기
- 가장 최근에 가입한 회원 조회하기
- 평균 나이보다 나이가 많은 회원 조회하기
- 이메일 주소를 기준으로 특정 회원 검색하기
- 원하는 조건 2개 이상을 함께 사용하여 회원 조회하기

````python
SELECT 
    CASE 
        WHEN age < 20 THEN '10대 이하'
        WHEN age < 30 THEN '20대'
        WHEN age < 40 THEN '30대'
        WHEN age < 50 THEN '40대'
        ELSE '50대 이상'
    END AS age_group,
    COUNT(*) AS member_count
FROM practice.members
GROUP BY age_group;
````

age_group	member_count
30대	2
20대	2

````python
SELECT *
FROM practice.members
ORDER BY joined_at DESC
LIMIT 1;
````

member_id	name	email	age	joined_at
4	정채원	chaewon@example.com	24	2026-08-15

````python
SELECT *
FROM practice.members
WHERE age > (
    SELECT AVG(age) 
    FROM practice.members
);
````
